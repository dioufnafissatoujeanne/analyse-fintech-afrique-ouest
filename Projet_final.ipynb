*QUESTION* 1:
Nombre d'utilisateurs par année




```python
utilisateurs_par_annee=df.groupby("Année")["Nombre_Utilisateurs"].sum()
```

```python
utilisateurs_par_annee
```

```python
plt.plot(utilisateurs_par_annee.index,
         utilisateurs_par_annee.values,
         marker="o")
```

Interprétation : Le graphique montre l'évolution du nombre d'utilisateurs des services bancaires mobiles entre 2000 et 2024. On observe une augmentation progressive du nombre d'utilisateurs au fil des années, ce qui traduit une adoption croissante des services financiers mobiles en Afrique de l'Ouest.

*QUESTION* 2:
Quels pays enregistrent les plus grands volumes de transactions ?






```python
volume_par_pays = df.groupby("Pays")["Volume_Transactions"].sum()

volume_par_pays
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    volume_par_pays.index,
    volume_par_pays.values,
    color="skyblue",
    edgecolor="black"
)

plt.title("Volume total des transactions par pays", fontsize=14)
plt.xlabel("Pays")
plt.ylabel("Volume des transactions")
plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

Interprétation

Le graphique compare le volume total des transactions entre les différents pays. Les pays ayant les barres les plus hautes sont ceux où les services financiers mobiles sont les plus utilisés.

*QUESTION* 3:Quels pays génèrent les valeurs de transactions les plus élevées ?

```python
valeur_par_pays = df.groupby("Pays")["Valeur_Transactions"].sum()

valeur_par_pays
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    valeur_par_pays.index,
    valeur_par_pays.values,
    color="green",
    edgecolor="black"
)

plt.title("Valeur totale des transactions par pays", fontsize=14)
plt.xlabel("Pays")
plt.ylabel("Valeur des transactions")
plt.xticks(rotation=45)

plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

Interprétation

Ce graphique montre les pays qui génèrent les montants financiers les plus importants via les services bancaires mobiles.

*Question* 4 : Quels types d'opérations sont les plus utilisés ?

```python
operations = df["Type_Operation"].value_counts()

operations
```

```python
plt.figure(figsize=(10,6))

plt.bar(
    operations.index,
    operations.values,
    color="orange",
    edgecolor="black"
)

plt.title("Types d'opérations les plus utilisés")
plt.xlabel("Type d'opération")
plt.ylabel("Nombre d'opérations")

plt.xticks(rotation=30)

plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```

*Question* 5 : Quels services Fintech sont les plus populaires ?

```python
services = df["Service_Fintech_Populaire"].value_counts()

services
```

```python
plt.figure(figsize=(8,8))

plt.pie(
    services,
    labels=services.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Services Fintech les plus populaires")

plt.show()
```

Interprétation

Le diagramme circulaire présente la part de chaque service fintech dans l'ensemble des services utilisés.

*Question* 6 : Existe-t-il une relation entre le taux de pénétration des fintechs et le nombre d'utilisateurs ?
Code

```python
plt.figure(figsize=(10,6))

plt.scatter(
    df["Taux_Penetration_Fintechs"],
    df["Nombre_Utilisateurs"],
    color="red"
)

plt.title("Taux de pénétration des Fintechs vs Nombre d'utilisateurs")
plt.xlabel("Taux de pénétration des Fintechs")
plt.ylabel("Nombre d'utilisateurs")

plt.grid(True)

plt.show()
```

Interprétation

Si les points montent globalement de gauche à droite, cela suggère qu'un taux de pénétration plus élevé est associé à un plus grand nombre d'utilisateurs.

*Question* 7 : Étudier la relation entre le revenu moyen et le nombre d'utilisateurs des services bancaires mobiles.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(df["Revenu_Moyen"], df["Nombre_Utilisateurs"], alpha=0.6)

plt.title("Relation entre le revenu moyen et le nombre d'utilisateurs")
plt.xlabel("Revenu moyen")
plt.ylabel("Nombre d'utilisateurs")
plt.grid(True)

plt.show()
```

Interprétation : Le graphique en nuage de points permet d'observer la relation entre le revenu moyen et le nombre d'utilisateurs des services bancaires mobiles. Les points sont répartis sur l'ensemble du graphique sans former une droite ou une courbe clairement identifiable. Cette dispersion suggère qu'il n'existe pas de relation linéaire forte entre ces deux variables. Ainsi, le revenu moyen ne semble pas, à lui seul, expliquer le nombre d'utilisateurs des services bancaires mobiles. D'autres facteurs, tels que l'accès à Internet, le taux de pénétration des Fintechs ou le niveau de bancarisation, peuvent également influencer l'adoption de ces services.
